```python
import cv2

# Open the webcam
cap = cv2.VideoCapture(0)

while True:

    # Read a frame from webcam
    ret, frame = cap.read()

    if not ret:
        break

    # Convert to grayscale
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Apply Gaussian Blur
    blur = cv2.GaussianBlur(gray, (5, 5), 0)

    # Detect edges
    edges = cv2.Canny(blur, 50, 150)

    # Find contours
    contours, _ = cv2.findContours(edges,
                                   cv2.RETR_EXTERNAL,
                                   cv2.CHAIN_APPROX_SIMPLE)

    # Detect shapes
    for cnt in contours:

        # Ignore very small objects
        if cv2.contourArea(cnt) > 1000:

            # Approximate contour
            approx = cv2.approxPolyDP(
                cnt,
                0.04 * cv2.arcLength(cnt, True),
                True
            )

            sides = len(approx)

            if sides == 3:
                shape = "Triangle"

            elif sides == 4:
                shape = "Rectangle"

            elif sides == 5:
                shape = "Pentagon"

            elif sides == 6:
                shape = "Hexagon"

            else:
                shape = "Circle"

            # Draw contour
            cv2.drawContours(frame, [approx], -1, (0, 255, 0), 2)

            # Display shape name
            x, y = approx[0][0]

            cv2.putText(frame,
                        shape,
                        (x, y - 10),
                        cv2.FONT_HERSHEY_SIMPLEX,
                        0.6,
                        (255, 0, 0),
                        2)

    # Show Edge Detection
    cv2.imshow("Edge Detection", edges)

    # Show Shape Detection
    cv2.imshow("Shape Detection", frame)

    # Press q to quit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Release webcam
cap.release()

# Close all windows
cv2.destroyAllWindows()
```
